## 1. Configuração do Ambiente Virtual
Criar um ambiente virtual usando venv para isolar as dependências do projeto.

````python
import os
import sys
import subprocess

print("Python Executável:", sys.executable)
print("Versão do Python:", sys.version)
print("Diretório de Trabalho:", os.getcwd())
print("\nPyPI ativo:", sys.prefix)
````

## 2. Verificação de Ferramentas Principais
Verificar pip, setuptools e wheel.

````python
import subprocess

# Verificar versões das ferramentas principais
result = subprocess.run([sys.executable, '-m', 'pip', '--version'], capture_output=True, text=True)
print("PIP Version:")
print(result.stdout)

# Verificar setuptools
result = subprocess.run([sys.executable, '-m', 'pip', 'show', 'setuptools'], capture_output=True, text=True)
print("\nSetuptools:")
print(result.stdout.split('\n')[0:2])

# Verificar wheel
result = subprocess.run([sys.executable, '-m', 'pip', 'show', 'wheel'], capture_output=True, text=True)
print("\nWheel:")
print(result.stdout.split('\n')[0:2])
````

## 3. Instalação de Bibliotecas de Ciência de Dados
Verificar e testar os pacotes fundamentais: NumPy, Pandas, SciPy e Scikit-learn.

````python
import numpy as np
import pandas as pd
import scipy
from sklearn import __version__ as sklearn_version

print("✓ NumPy:", np.__version__)
print("✓ Pandas:", pd.__version__)
print("✓ SciPy:", scipy.__version__)
print("✓ Scikit-learn:", sklearn_version)

# Teste rápido de funcionalidade
arr = np.array([1, 2, 3, 4, 5])
df = pd.DataFrame({'valores': arr, 'quadrado': arr**2})
print("\nTeste de Funcionalidade:")
print(df)
````

## 4. Instalação de Ferramentas de Visualização
Verificar as bibliotecas de visualização: Matplotlib, Seaborn e Plotly.

````python
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
import plotly
import plotly.express as px

print("✓ Matplotlib:", matplotlib.__version__)
print("✓ Seaborn:", sns.__version__)
print("✓ Plotly:", plotly.__version__)

# Configurar matplotlib para Jupyter
%matplotlib inline

# Teste de visualização simples
fig, ax = plt.subplots(figsize=(10, 5))
x = np.linspace(0, 10, 100)
y = np.sin(x)
ax.plot(x, y, 'b-', label='sin(x)')
ax.plot(x, np.cos(x), 'r-', label='cos(x)')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Teste de Visualização com Matplotlib')
ax.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

## 5. Relatório Completo de Verificação de Instalações
Exibir um resumo detalhado de todas as bibliotecas instaladas.

````python
import pkg_resources

# Lista de pacotes essenciais
pacotes_essenciais = [
    'numpy', 'pandas', 'scipy', 'scikit-learn',
    'matplotlib', 'seaborn', 'plotly',
    'jupyter', 'jupyterlab', 'ipython',
    'requests', 'beautifulsoup4'
]

print("=" * 60)
print("RELATÓRIO DE INSTALAÇÃO - PACOTES ESSENCIAIS")
print("=" * 60)

versoes = {}
for pacote in pacotes_essenciais:
    try:
        version = pkg_resources.get_distribution(pacote).version
        versoes[pacote] = version
        print(f"✓ {pacote:20} v{version}")
    except:
        print(f"✗ {pacote:20} NÃO INSTALADO")

print("=" * 60)
print(f"Total de pacotes verificados: {len(versoes)}/{len(pacotes_essenciais)}")
````
